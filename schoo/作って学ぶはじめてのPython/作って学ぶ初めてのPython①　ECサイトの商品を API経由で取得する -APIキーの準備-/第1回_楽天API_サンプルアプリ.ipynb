```python
import os
from pathlib import Path

import requests
from dotenv import load_dotenv
```

```python
# ===== 環境変数の読み込み =====
# .pyを使うとき
# env_path = Path(__file__).parent
# .ipynbを使うとき
env_path = Path.cwd().parent / ".env"
load_dotenv(dotenv_path=env_path)
```

```python
# ===== 定数定義 =====
APP_ID = os.getenv("RAKUTEN_APP_ID")  # ここに取得したアプリケーションIDを入力
QUERY = "水 2L"  # 検索したいキーワード
```

```python
url = "https://app.rakuten.co.jp/services/api/IchibaItem/Search/20220601"
params = {
    "applicationId": APP_ID,
    "format": "json",
    "keyword": QUERY
}

response = requests.get(url, params=params)

if response.status_code == 200:
    json_data = response.json()
    items = json_data["Items"]
    print(items)
    for item in items:
        print(item["Item"])
else:
    print(f"Error: {response.status_code}")
```

```python
# ===== 定数定義 =====
APP_ID = os.getenv("RAKUTEN_APP_ID")  # ここに取得したアプリケーションIDを入力
QUERY = "段ボール"  # 検索したいキーワード
```

```python
url = "https://app.rakuten.co.jp/services/api/IchibaItem/Search/20220601"
params = {"applicationId": APP_ID, "format": "json", "keyword": QUERY}

response = requests.get(url, params=params)

if response.status_code == 200:
    json_data = response.json()
    items = json_data["Items"]

    for item in items:
        print(f"商品名: {item['Item']['itemName']}")
        print(f"価格: {item['Item']['itemPrice']}円")
        print(f"URL: {item['Item']['itemUrl']}")
        print("-" * 50)
else:
    print(f"Error: {response.status_code}")
```
